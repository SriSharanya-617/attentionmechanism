Attention Mechanism

```python
import numpy as np
from tensorflow.keras.models import Model

from tensorflow.keras.layers import(
Input,
LSTM,
Embedding,
Dense,
Attention,
Concatenate

)


from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
english_sentences = [
    "hi",
    "how are you",
    "i am good",
    "good morning",
    "good evening",
    "good night",
    "i love deep learning",
    "i love ai",
    "what is your name",
    "my name is google assistant",
    "i like to learn",
    "tell me about yourself",
    "where are you from",
    "i am learning english",
    "can you help me",
    "yes i can help you",
    "thank you",
    "you are welcome",
    "how old are you",
    "i am a computer program"
]

french_sentences = [
    "start salut end",
    "start comment allez-vous end",
    "start je vais bien end",
    "start bonjour end",
    "start bonsoir end",
    "start bonne nuit end",
    "start j'aime l'apprentissage profond end",
    "start j'aime l'IA end",
    "start quel est votre nom end",
    "start mon nom est assistant google end",
    "start j'aime apprendre end",
    "start parlez-moi de vous end",
    "start d'où venez-vous end",
    "start j'apprends l'anglais end",
    "start pouvez-vous m'aider end",
    "start oui je peux vous aider end",
    "start merci end",
    "start de rien end",
    "start quel âge avez-vous end",
    "start je suis un programme informatique end"
]

print("English Sentences")
print(english_sentences)
print("French Sentences")
print(french_sentences)
```

Tokenization

1.   neural networks understand numbers,not text




```python
eng_tokenizer = Tokenizer()
eng_tokenizer.fit_on_texts(english_sentences)

fra_tokenizer = Tokenizer()
fra_tokenizer.fit_on_texts(french_sentences)

print("English Vocabulary:")
print(eng_tokenizer.word_index)
print("French Vocabulary:")
print(fra_tokenizer.word_index)
```

Convert Sentences into Sequences

```python
encoder_input = eng_tokenizer.texts_to_sequences(english_sentences)
decoder_input = fra_tokenizer.texts_to_sequences(french_sentences)

print("English Sequences:")
print(encoder_input)
print("French Sequences:")
print(decoder_input)
```

padding

---



*   padding makes all sentences of equal length



```python
encoder_input = pad_sequences(
    encoder_input,
    padding='post'
)

decoder_input = pad_sequences(
    decoder_input,
    padding='post'
)

print("English Sequences:")
print(encoder_input)
print("French Sequences:")
print(decoder_input)
```

Build encoder

* encoder generated hidden states for every word



```python
encoder_inputs = Input(shape=(None,))
encoder_embedding = Embedding(
    input_dim=len(eng_tokenizer.word_index) + 1,
    output_dim=64
)(encoder_inputs)
encoder_outputs, state_h, state_c = LSTM(
    units=64,
    return_sequences=True,
    return_state=True
)(encoder_embedding)
print("encoder created successfully")
```

Important Understanding

1.encoder_outputs:: Hidden states for all words
2.state_h : final hidden state
3.state_c :final cell state



Build decoder

```python
decoder_inputs = Input(shape=(None,))
decoder_embedding_layer = Embedding(
    input_dim=len(fra_tokenizer.word_index) + 1,
    output_dim=64
)
decoder_embedding_output = decoder_embedding_layer(decoder_inputs)

decoder_outputs, _, _ = LSTM(
    units=64,
    return_sequences=True,
    return_state=True
)(decoder_embedding_output, initial_state=[state_h, state_c])

print("decoder Created Successfully")
```

Attention layer

1.   decode outputs with enocde woutput
and dynamically focus on important words




```python
attention=Attention()
attention_output=attention([decoder_outputs, encoder_outputs])

print("Attention Layer Created Successfully")
```

Combine Decoder + Attention Output

```python
decoder_combined=Concatenate(axis =- 1)([decoder_outputs,attention_output])

print("Decoder + Attention Combined Successfully")
```

Final Prediction Layer

```python
output = Dense(
    len(fra_tokenizer.word_index) + 1,
    activation='softmax'
)(decoder_combined)

print("Prediction Layer Created Successfully")
```

build model

```python
model=Model([encoder_inputs, decoder_inputs],output)

print("Model Created Successfully")
model.summary()
```

compile model

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

train model

```python
decoder_target = np.expand_dims(decoder_input, -1)

model.fit(
    [encoder_input, decoder_input],
    decoder_target,
    epochs=50, # Increased epochs for better learning with larger dataset
    batch_size=2
)
```

## Prediction

```python
# Inference Encoder Model
encoder_model_inference = Model(encoder_inputs, [encoder_outputs, state_h, state_c])
print("Inference Encoder Model Created Successfully")

# Inference Decoder Model
decoder_inference_inputs = Input(shape=(None,)) # New input for inference decoder
decoder_state_input_h = Input(shape=(64,))
decoder_state_input_c = Input(shape=(64,))
encoder_outputs_input = Input(shape=(None, 64))

decoder_outputs_inference, state_h_inference, state_c_inference = LSTM(
    units=64,
    return_sequences=True,
    return_state=True
)(decoder_embedding_layer(decoder_inference_inputs), initial_state=[decoder_state_input_h, decoder_state_input_c])

attention_output_inference = Attention()([decoder_outputs_inference, encoder_outputs_input])
decoder_combined_inference = Concatenate(axis=-1)([decoder_outputs_inference, attention_output_inference])

output_tokens_inference = Dense(
    len(fra_tokenizer.word_index) + 1,
    activation='softmax'
)(decoder_combined_inference)

decoder_model_inference = Model(
    [decoder_inference_inputs, encoder_outputs_input, decoder_state_input_h, decoder_state_input_c],
    [output_tokens_inference, state_h_inference, state_c_inference]
)
print("Inference Decoder Model Created Successfully")

# Create a reverse word index for French for faster lookup
idx_to_word_fra = {idx: word for word, idx in fra_tokenizer.word_index.items()}

# Define a maximum sequence length for decoding based on the training data's max token length
max_decoder_sequence_length = decoder_input.shape[1] # This represents the max number of tokens

def predict_sequence(input_seq):
    # Encode the input as state vectors
    encoder_out, h, c = encoder_model_inference.predict(input_seq)

    # Generate empty target sequence of length 1 with the start token
    target_seq = np.zeros((1, 1))
    target_seq[0, 0] = fra_tokenizer.word_index['start']

    decoded_sentence = []
    # Loop for a maximum number of steps to prevent infinite loops
    for _ in range(max_decoder_sequence_length): # Limit decoding steps to max training sequence length
        output_tokens, h, c = decoder_model_inference.predict(
            [target_seq, encoder_out, h, c])

        # Sample a token (greedy search)
        sampled_token_index = np.argmax(output_tokens[0, -1, :])

        # Convert index to word. Use '' for unknown/padding (index 0 might be padding)
        sampled_word = idx_to_word_fra.get(sampled_token_index, '')

        if sampled_word == 'end':
            break # Stop if 'end' token is predicted
        elif sampled_word != '': # Only append actual words (not padding or empty strings)
            decoded_sentence.append(sampled_word)

        # Update the target sequence (of length 1) for the next step.
        target_seq = np.zeros((1, 1))
        target_seq[0, 0] = sampled_token_index

    return ' '.join(decoded_sentence).strip() # Join the collected words

print("Prediction function defined")

for i in range(len(english_sentences)):
    input_sentence = english_sentences[i]
    input_sequence = eng_tokenizer.texts_to_sequences([input_sentence])
    input_sequence_padded = pad_sequences(input_sequence, padding='post', maxlen=encoder_input.shape[1])
    translated_sentence = predict_sequence(input_sequence_padded)
    print(f"English: {input_sentence} => French: {translated_sentence}")
```
